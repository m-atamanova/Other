В идеале, мы хотим идти по порядку - так затраченное время будет наименьшим (не важно снизу вверх или сверху вниз), поэтому проверим лежит ли этаж убегающего сотрудника по этому пути наименьшего следования.

- Для этого разница между его этажом и первым(последним) должна быть меньше времени его пребывания (чтобы мы успели до него дойти). Если он нас легчайше дожидается, то суммарное затраченное время на обход всех коллег равно разнице между первым и последним этажом.

```python
if floor_colleague - floors[0] < time_before_go:
    print(floors[-1] - floors[0])
elif floors[-1] - floor_colleague  < time_before_go:
    print(floors[-1] - floors[0])
else:
    #floors.append(floor_colleague)
    #new_floors = sorted(floors) #сортируем по возрастанию
    if abs(floor_colleague - floors[0]) < abs(floor_colleague - floors[-1]):
        print(abs(floor_colleague - floors[0])  + (floors[-1] - floors[0]))
    else:
        print(abs(floor_colleague - floors[-1])  + (floors[-1] - floors[0]))
```

Примем на вход переменные

```python
n, k = map(int, input().split()) # n - кол-во чисел, k-число операций
all_numbers = []
for i in range(n):
    a = int(input())
    all_numbers.append(a)
print(all_numbers)
```

- проверяем число разрядов
- сортируем по отдельным спискам согласно разрядности и внутри них - по возрастанию
- если списки не пустые, то добавляем в финальный список списков, который сортируем по убыванию

```python
digit_9 = []
digit_8 = []
digit_7 = []
digit_6 = []
digit_5 = []
digit_4 = []
digit_3 = []
digit_2 = []
digit_1 = []

for i in all_numbers:
    digit = len(str(i))
    if digit == 9:
        digit_9.append(i)
    elif digit == 8:
        digit_8.append(i)
    elif digit == 7:
        digit_7.append(i)
    elif digit ==6:
        digit_6.append(i)
    elif digit == 5:
        digit_5.append(i)
    elif digit == 4:
        digit_4.append(i)
    elif digit == 3:
        digit_3.append(i)
    elif digit == 2:
        digit_2.append(i)
    elif digit == 1:
        digit_1.append(i)

digit_9.sort()
digit_8.sort()
digit_7.sort()
digit_6.sort()
digit_5.sort()
digit_4.sort()
digit_3.sort()
digit_2.sort()
digit_1.sort()

all_digits = []
if digit_9:
    all_digits.append(digit_9)
if digit_8:
    all_digits.append(digit_8)
if digit_7:
    all_digits.append(digit_7)
if digit_6:
    all_digits.append(digit_6)
if digit_5:
    all_digits.append(digit_5)
if digit_4:
    all_digits.append(digit_4)
if digit_3:
    all_digits.append(digit_3)
if digit_2:
    all_digits.append(digit_2)
if digit_1:
    all_digits.append(digit_1)

print(all_digits)
```

Наибольший вклад в разницу между старой и новой суммой цифр вносят **большие** разряды, поэтому менять надо именно их в первую очередь - "9" на месте сотен даст больше разницы чем "9" на месте единиц. Поэтому мы отсортировали по разрядам и возрастанию.

Проверяем два условия и проводим работу пока k остается > 0 (не закончится число корректировок):
- идем по спискам списка, затем проьегаем в них каждый элемент
1. - если он не начинается на 9, то меняем первую цифру на 9
   - как только k == 0 выходим из подцикла и затем цикла
   - считаем разницу
2. - чтобы учесть случай, когда число изменений k остается больше, чем вообще возможно поменять цифр, вводим flag = False
   - если хотя бы одна переменная меняется, то он становится True.
   - если флаг остается False - программа выходит из цикла while переходит к следующему блоку (печатает ноль, если ничего не было изменено; печатает разницу если что-то менялось до выхода)

```python
while k > 0:
    flag = False # проверяет изменили ли мы хоть 1 число
    for i in all_digits:
        for j in range(len(i)):
            first_digit = int(str(i[j])[0])
            if first_digit != 9:
                i[j] = int('9' + str(i[j])[1:])
                flag = True
                k -= 1
                if k == 0:
                    break #выход из подцикла
        if k == 0: # выход из цикла
            break
    if flag:
        continue
    else:
        break


```
